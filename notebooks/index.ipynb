# Welcome to the 2007 Llama Caravan project geodata

This notebook uses python to display geodata derived from our 2007 journey with a llama caravan in southern Peru.

## Setup

```python
!pip install geopandas folium fiona
import geopandas as gpd
import folium
import fiona
```

```python

# Enable KML driver
fiona.drvsupport.supported_drivers['KML'] = 'rw'

# Read KML files
url_points = 'https://raw.githubusercontent.com/ntrip6/caravan2007/main/data/caravan_camps.kml'
url_lines = 'https://raw.githubusercontent.com/ntrip6/caravan2007/main/data/caravan_route_simplified.kml'

gdf_points = gpd.read_file(url_points, driver='KML')
gdf_lines = gpd.read_file(url_lines, driver='KML')

# Center map on Peru data
center = [-14.85, -72.96]

# Create map with OpenTopoMap as default
m = folium.Map(
    location=center,
    zoom_start=10,
    tiles='https://tile.opentopomap.org/{z}/{x}/{y}.png',
    attr='OpenTopoMap'
)

# Add ESRI Terrain basemap
folium.TileLayer(
    tiles='https://server.arcgisonline.com/ArcGIS/rest/services/World_Shaded_Relief/MapServer/tile/{z}/{y}/{x}',
    attr='ESRI World Shaded Relief',
    name='ESRI Terrain'
).add_to(m)

# Add Google Roads basemap
folium.TileLayer(
    tiles='https://mt1.google.com/vt/lyrs=m&x={x}&y={y}&z={z}',
    attr='Google Roads',
    name='Google Roads'
).add_to(m)

# Add points layer
folium.GeoJson(
    gdf_points,
    name='Caravan Camps',
    tooltip=folium.GeoJsonTooltip(fields=['Name', 'Description'])
).add_to(m)

# Add lines layer
folium.GeoJson(
    gdf_lines,
    name='Caravan Route',
    style_function=lambda x: {'color': 'red', 'weight': 2}
).add_to(m)

# Add layer control
folium.LayerControl().add_to(m)

m
```
